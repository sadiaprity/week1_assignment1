## 2. The question: decision, action, cost of a wrong call

**Decision improved:** which content signals a reviewer should pay attention to when deciding
whether a page is worth reviewing, refreshing, or leaving alone.

**Who acts on it:** a content/SEO reviewer with limited time, who needs to know *what to look at*
before opening a page.

**Output:** a short signal report — which signals (e.g. impressions, word count, CTR, position,
freshness) are associated with higher or lower performance, with the size of the association and
charts to show it.

**Cost of a wrong call:** if the report says a signal matters when it doesn't, reviewers waste
time chasing noise. If it says a signal doesn't matter when it does, real opportunities get missed.
Because this is observational (not experimental), a wrong call is "we chased a weak pattern" — not
"we broke a live system." The damage is wasted review time, not user harm.

```python
import pandas as pd
import numpy as np

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")
print("rows:", len(df), " columns:", len(df.columns))

# Number 1: how common is decline in the starter slice?
print("\n1) Share of pages labelled declining (trend_direction == 'down'):")
print(round((df["trend_direction"] == "down").mean(), 3))

# Number 2: do visible pages actually differ in clicks by position tier?
# (position tier is a simple bucket; we just compare medians)
df["pos_tier"] = pd.cut(df["avg_position"], bins=[0, 3, 10, 20, 100],
                        labels=["1-3", "4-10", "11-20", "21+"])
print("\n2) Median 90d clicks by average-position tier:")
print(df.groupby("pos_tier", observed=True)["clicks_90d"].median())

# Number 3: does word count relate to visibility? (weak/strong, we just show it)
print("\n3) Correlation between word_count and impressions_90d:")
print(round(df[["word_count", "impressions_90d"]].corr().iloc[0, 1], 3))

print("\nWhy this makes Lane 1 worth it: decline is common enough to study, "
      "position clearly separates pages by clicks, and word count has a measurable link to visibility.")
```

## 4. Careful words: what I can and can't claim

**What this work CAN say (and will say, carefully):**
- We observed an association between signal X and outcome Y in the starter slice.
- Pages with higher X tended to have higher/lower Y.
- The size of the effect was small / moderate / large in this data.
- This is decision-support: it helps a reviewer prioritize, not guarantees.

**What this work will NEVER claim:**
- That a signal causes better or worse performance (no experiment, no causal design).
- That we have identified a Google ranking factor.
- That refreshing a page will cause it to recover.
- That the pattern holds on the full ~79M-row warehouse — this is the 30k starter slice only.

Every claim will use words like observed, measured, associated, directional, decision-support —
never proves, causes, guarantees, predicts.
